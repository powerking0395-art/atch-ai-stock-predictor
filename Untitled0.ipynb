```python
# ============================================
# Step 2 - Download ATCH Stock Data
# ============================================

TICKER = "ATCH"

stock = yf.download(
    TICKER,
    start="2020-01-01",
    end=pd.Timestamp.today().strftime("%Y-%m-%d"),
    auto_adjust=False
)

stock.head()
```

```python
print("데이터 크기:", stock.shape)

print("\n컬럼:")
print(stock.columns)

print("\n최근 데이터:")
display(stock.tail())
```

```python
# ============================================
# Step 3 - Clean Stock Data
# ============================================

stock = stock.copy()

# MultiIndex 컬럼 처리
if isinstance(stock.columns, pd.MultiIndex):
    stock.columns = stock.columns.get_level_values(0)

# 날짜를 컬럼으로 변환
stock.reset_index(inplace=True)

# 필요한 컬럼만 사용
stock = stock[
    [
        "Date",
        "Open",
        "High",
        "Low",
        "Close",
        "Volume"
    ]
].copy()

# 결측치 제거
stock.dropna(inplace=True)

# 날짜 정렬
stock.sort_values("Date", inplace=True)

# 인덱스 초기화
stock.reset_index(drop=True, inplace=True)

display(stock.head())
display(stock.tail())
```

```python
# ============================================
# Step 4 - Technical Indicators
# ============================================

# 일간 수익률
stock["Return_1D"] = stock["Close"].pct_change()

# 이동평균
stock["MA5"] = stock["Close"].rolling(5).mean()
stock["MA20"] = stock["Close"].rolling(20).mean()
stock["MA60"] = stock["Close"].rolling(60).mean()

# 변동성
stock["Volatility_20D"] = (
    stock["Return_1D"]
    .rolling(20)
    .std()
)

# 거래량 이동평균
stock["Volume_MA20"] = (
    stock["Volume"]
    .rolling(20)
    .mean()
)

# 거래량 변화율
stock["Volume_Change"] = (
    stock["Volume"].pct_change()
)

display(stock.tail())
```

```python
# ============================================
# Step 5 - RSI
# ============================================

delta = stock["Close"].diff()

gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)

avg_gain = gain.rolling(14).mean()
avg_loss = loss.rolling(14).mean()

rs = avg_gain / avg_loss

stock["RSI14"] = 100 - (100 / (1 + rs))

display(
    stock[
        [
            "Date",
            "Close",
            "Volume",
            "MA5",
            "MA20",
            "MA60",
            "RSI14"
        ]
    ].tail(10)
)
```

```python
# ============================================
# Step 6 - Market Data
# ============================================

market = yf.download(
    ["SPY", "QQQ"],
    start="2020-01-01",
    end=pd.Timestamp.today().strftime("%Y-%m-%d"),
    auto_adjust=False
)

market.head()
```

```python
# SPY
spy = market["Close"]["SPY"].pct_change()
spy.name = "SPY_Return"

# QQQ
qqq = market["Close"]["QQQ"].pct_change()
qqq.name = "QQQ_Return"

# ATCH와 날짜 기준 병합
stock = stock.merge(
    spy,
    left_on="Date",
    right_index=True,
    how="left"
)

stock = stock.merge(
    qqq,
    left_on="Date",
    right_index=True,
    how="left"
)

display(stock.tail())
```

```python
# ============================================
# Step 7 - Target Variable
# ============================================

# 다음 거래일 종가
stock["Next_Close"] = stock["Close"].shift(-1)

# 다음 거래일 수익률
stock["Next_Return"] = (
    stock["Next_Close"] / stock["Close"] - 1
)

# 상승 = 1
# 하락 또는 동일 = 0
stock["Target"] = (
    stock["Next_Return"] > 0
).astype(int)

display(
    stock[
        [
            "Date",
            "Close",
            "Next_Close",
            "Next_Return",
            "Target"
        ]
    ].tail(10)
)
```

```python
# ============================================
# Step 9 - SEC Filing Data
# ============================================

import requests
import pandas as pd
import numpy as np

SEC_CIK = "0001963088"

headers = {
    "User-Agent": "ATCH AI Stock Prediction Project research@example.com"
}

url = f"https://data.sec.gov/submissions/CIK{SEC_CIK}.json"

response = requests.get(url, headers=headers)

print("HTTP Status:", response.status_code)

sec_data = response.json()

print("회사명:", sec_data["name"])
print("티커:", sec_data["tickers"])
```

```python
# ============================================
# Step 10 - Filing History
# ============================================

recent = sec_data["filings"]["recent"]

filings = pd.DataFrame(recent)

filings = filings[
    [
        "filingDate",
        "reportDate",
        "form",
        "accessionNumber",
        "primaryDocument"
    ]
].copy()

filings["filingDate"] = pd.to_datetime(
    filings["filingDate"]
)

filings["reportDate"] = pd.to_datetime(
    filings["reportDate"],
    errors="coerce"
)

filings.sort_values(
    "filingDate",
    ascending=False,
    inplace=True
)

display(filings.head(30))
```

```python
# ============================================
# Step 11 - Important Filings
# ============================================

important_forms = [
    "10-K",
    "10-Q",
    "8-K",
    "424B2",
    "424B3",
    "424B4",
    "DEF 14A",
    "S-1",
    "S-3",
    "13D",
    "13G"
]

important_filings = filings[
    filings["form"].isin(important_forms)
].copy()

print(
    "중요 공시 개수:",
    len(important_filings)
)

display(
    important_filings.head(30)
)
```

```python
# ============================================
# Step 12 - Daily Filing Count
# ============================================

filing_daily = (
    important_filings
    .groupby("filingDate")
    .size()
    .reset_index(name="Filing_Count")
)

filing_daily.head()
```

```python
stock["Date"] = pd.to_datetime(stock["Date"])

stock = stock.merge(
    filing_daily,
    left_on="Date",
    right_on="filingDate",
    how="left"
)

stock.drop(
    columns=["filingDate"],
    inplace=True
)

stock["Filing_Count"] = (
    stock["Filing_Count"]
    .fillna(0)
)

display(
    stock[
        ["Date", "Close", "Filing_Count"]
    ].tail(20)
)
```

```python
# ============================================
# Step 13 - Filing Type Features
# ============================================

for form in ["8-K", "10-Q", "10-K", "S-1", "S-3", "DEF 14A"]:

    temp = (
        important_filings[
            important_filings["form"] == form
        ]
        .groupby("filingDate")
        .size()
        .reset_index(name=f"Filing_{form.replace('-', '_')}")
    )

    stock = stock.merge(
        temp,
        left_on="Date",
        right_on="filingDate",
        how="left"
    )

    stock.drop(
        columns=["filingDate"],
        inplace=True
    )

    stock[f"Filing_{form.replace('-', '_')}"] = (
        stock[f"Filing_{form.replace('-', '_')}"]
        .fillna(0)
    )

display(stock.tail())
```

```python
# ============================================
# Step 14 - Filing Event Categories
# ============================================

def classify_filing(form):

    if form == "10-K":
        return "ANNUAL_REPORT"

    elif form == "10-Q":
        return "QUARTERLY_REPORT"

    elif form == "8-K":
        return "CURRENT_EVENT"

    elif form in ["S-1", "S-3"]:
        return "SECURITIES"

    elif form == "DEF 14A":
        return "SHAREHOLDER_MEETING"

    else:
        return "OTHER"


important_filings["Event_Type"] = (
    important_filings["form"]
    .apply(classify_filing)
)

display(
    important_filings[
        [
            "filingDate",
            "form",
            "Event_Type"
        ]
    ].head(30)
)
```

```python
# ============================================
# Step 20 - Combine Financial Data
# ============================================

import pandas as pd

# 1. 이전 단계에서 누락된 financial_data 딕셔너리 생성 및 데이터 추출 진행
financial_data = {}
for feature, concept in selected_concepts.items():
    try:
        financial_data[feature] = extract_financial_fact(concept)
    except Exception as e:
        print(f"{feature} ({concept}) 추출 중 오류 발생: {e}")

# 2. 재무 데이터 결합 작업 진행
financial_records = []

for feature, df in financial_data.items():
    temp = df.copy()
    temp["Feature"] = feature
    financial_records.append(temp)

if financial_records:
    financial_df = pd.concat(
        financial_records,
        ignore_index=True
    )

    financial_df.sort_values(
        "filed",
        inplace=True
    )

    financial_df.reset_index(
        drop=True,
        inplace=True
    )

    print(
        "재무 데이터 크기:",
        financial_df.shape
    )

    display(
        financial_df[
            [
                "Feature",
                "start",
                "end",
                "filed",
                "val",
                "form"
            ]
        ].tail(30)
    )
else:
    print("결합할 재무 데이터가 존재하지 않습니다.")
```

```python
# ============================================
# Step 33 - Merge Shares Outstanding
# ============================================

# 만약 앞 단계에서 shares_df가 생성되지 않았다면 대안 개념 검색
if 'shares_df' not in globals():
    print("shares_df가 존재하지 않아 대안 개념(CommonStockSharesOutstanding) 조회를 시도합니다.")
    alt_shares_concept = "CommonStockSharesOutstanding"
    if alt_shares_concept in us_gaap:
        shares_df = extract_financial_fact(alt_shares_concept)
        shares_df = shares_df[["end", "filed", "val", "form"]].copy()
        shares_df.rename(
            columns={
                "end": "Shares_Date",
                "filed": "Available_Date",
                "val": "Shares_Outstanding"
            },
            inplace=True
        )
        shares_df["Shares_Date"] = pd.to_datetime(shares_df["Shares_Date"])
        shares_df["Available_Date"] = pd.to_datetime(shares_df["Available_Date"])
        shares_df.sort_values("Available_Date", inplace=True)
    else:
        shares_df = None

stock = stock.sort_values("Date")

if shares_df is not  None:
    shares_df = shares_df.sort_values("Available_Date")

    stock = pd.merge_asof(
        stock,
        shares_df[
            [
                "Available_Date",
                "Shares_Outstanding"
            ]
        ],
        left_on="Date",
        right_on="Available_Date",
        direction="backward"
    )

    stock.drop(
        columns=["Available_Date"],
        inplace=True
    )

    display(
        stock[
            [
                "Date",
                "Close",
                "Shares_Outstanding"
            ]
        ].tail(20)
    )
else:
    print("발행주식수(Shares Outstanding) 데이터를 찾을 수 없어 병합을 스킵합니다. 'Shares_Outstanding' 컬럼을 NaN으로 추가합니다.")
    stock["Shares_Outstanding"] = np.nan
```

```python
# ============================================
# Step 34 - Market Capitalization
# ============================================

stock["Market_Cap"] = (
    stock["Close"] *
    stock["Shares_Outstanding"]
)

display(
    stock[
        [
            "Date",
            "Close",
            "Shares_Outstanding",
            "Market_Cap"
        ]
    ].tail(20)
)
```

```python
# ============================================
# Step 35 - Prepare Financial Data
# ============================================

financial_pivot = financial_df.copy()

financial_pivot = financial_pivot[
    [
        "Feature",
        "filed",
        "end",
        "val",
        "form"
    ]
].copy()

financial_pivot["filed"] = pd.to_datetime(
    financial_pivot["filed"]
)

financial_pivot["end"] = pd.to_datetime(
    financial_pivot["end"]
)

# 필요한 재무항목만 선택
wanted_features = [
    "Revenue",
    "Net_Income",
    "Assets",
    "Liabilities",
    "Stockholders_Equity",
    "Cash",
    "Operating_Cash_Flow"
]

financial_pivot = financial_pivot[
    financial_pivot["Feature"].isin(wanted_features)
].copy()

display(financial_pivot.tail(30))
```

```python
# ============================================
# Step 36 - Latest Available Financial Data
# ============================================

financial_pivot = financial_pivot.sort_values(
    ["Feature", "filed", "end"]
)

# 같은 Feature + 같은 공시일 + 같은 종료일이
# 여러 개 존재하는 경우 마지막 값 사용
financial_pivot = (
    financial_pivot
    .drop_duplicates(
        subset=["Feature", "filed", "end"],
        keep="last"
    )
)

# Feature를 column으로 변환
financial_wide = financial_pivot.pivot_table(
    index="filed",
    columns="Feature",
    values="val",
    aggfunc="last"
).reset_index()

financial_wide.rename(
    columns={"filed": "Financial_Available_Date"},
    inplace=True
)

financial_wide.sort_values(
    "Financial_Available_Date",
    inplace=True
)

display(financial_wide.tail(20))
```

```python
# ============================================
# Step 37 - Merge Financial Data
# ============================================

stock = stock.sort_values("Date")
financial_wide = financial_wide.sort_values(
    "Financial_Available_Date"
)

stock = pd.merge_asof(
    stock,
    financial_wide,
    left_on="Date",
    right_on="Financial_Available_Date",
    direction="backward"
)

stock.drop(
    columns=["Financial_Available_Date"],
    inplace=True
)

display(
    stock[
        [
            "Date",
            "Close",
            "Revenue",
            "Net_Income",
            "Assets",
            "Liabilities",
            "Stockholders_Equity",
            "Cash"
        ]
    ].tail(20)
)
```

```python
# ============================================
# Step 38 - Price to Sales
# ============================================

stock["PS_Ratio"] = (
    stock["Market_Cap"] /
    stock["Revenue"]
)

display(
    stock[
        [
            "Date",
            "Market_Cap",
            "Revenue",
            "PS_Ratio"
        ]
    ].tail(20)
)
```

```python
# ============================================
# Step 39 - Price to Book
# ============================================

stock["PB_Ratio"] = (
    stock["Market_Cap"] /
    stock["Stockholders_Equity"]
)

display(
    stock[
        [
            "Date",
            "Market_Cap",
            "Stockholders_Equity",
            "PB_Ratio"
        ]
    ].tail(20)
)
```

```python
# ============================================
# Step 40 - Debt Ratio
# ============================================

stock["Debt_Ratio"] = (
    stock["Liabilities"] /
    stock["Assets"]
)

display(
    stock[
        [
            "Date",
            "Assets",
            "Liabilities",
            "Debt_Ratio"
        ]
    ].tail(20)
)
```

```python
# ============================================
# Step 41 - Cash to Liabilities
# ============================================

stock["Cash_to_Liabilities"] = (
    stock["Cash"] /
    stock["Liabilities"]
)

display(
    stock[
        [
            "Date",
            "Cash",
            "Liabilities",
            "Cash_to_Liabilities"
        ]
    ].tail(20)
)
```

```python
# ============================================
# Step 42 - ROA
# ============================================

stock["ROA"] = (
    stock["Net_Income"] /
    stock["Assets"]
)

display(
    stock[
        [
            "Date",
            "Net_Income",
            "Assets",
            "ROA"
        ]
    ].tail(20)
)
```

```python
# ============================================
# Step 43 - ROE
# ============================================

stock["ROE"] = (
    stock["Net_Income"] /
    stock["Stockholders_Equity"]
)

stock.loc[
    stock["Stockholders_Equity"] == 0,
    "ROE"
] = np.nan

display(
    stock[
        [
            "Date",
            "Net_Income",
            "Stockholders_Equity",
            "ROE"
        ]
    ].tail(20)
)
```

```python
# ============================================
# Step 46 - ATCH Earnings Events
# ============================================

earnings_events = pd.DataFrame({
    "Event_Date": pd.to_datetime([
        "2024-05-28",
        "2024-11-15",
        "2025-02-19",
        "2025-05-16",
        "2025-09-29",
        "2025-11-14",
        "2026-02-13",
        "2026-05-13",
        "2026-09-23"
    ]),

    "Event_Type": [
        "EARNINGS",
        "EARNINGS",
        "EARNINGS",
        "EARNINGS",
        "EARNINGS",
        "EARNINGS",
        "EARNINGS",
        "EARNINGS",
        "EARNINGS"
    ]
})

earnings_events = earnings_events.sort_values("Event_Date")

display(earnings_events)
```

```python
# ============================================
# Step 48 - Days To Earnings
# ============================================

# Date 컬럼을 datetime 형식으로 확실히 변환
stock["Date"] = pd.to_datetime(stock["Date"])
earnings_events["Event_Date"] = pd.to_datetime(earnings_events["Event_Date"])

# 각 주가 데이터 날짜 기준으로 다음(가장 가까운 미래) 실적 발표일을 계산하는 함수 정의
def get_next_earnings_date(current_date):
    future_events = earnings_events[earnings_events["Event_Date"] >= current_date]
    if not future_events.empty:
        return future_events["Event_Date"].min()
    return pd.NaT

# Next_Earnings_Date 컬럼 생성
stock["Next_Earnings_Date"] = stock["Date"].apply(get_next_earnings_date)

# Days_To_Earnings 계산
stock["Days_To_Earnings"] = (
    stock["Next_Earnings_Date"] -
    stock["Date"]
).dt.days

display(
    stock[
        [
            "Date",
            "Close",
            "Next_Earnings_Date",
            "Days_To_Earnings"
        ]
    ].tail(30)
)
```

```python
# ============================================
# Step 49 - Earnings Event Window
# ============================================

stock["Earnings_Within_5D"] = (
    stock["Days_To_Earnings"].between(0, 5)
).astype(int)

stock["Earnings_Within_10D"] = (
    stock["Days_To_Earnings"].between(0, 10)
).astype(int)

stock["Earnings_Within_20D"] = (
    stock["Days_To_Earnings"].between(0, 20)
).astype(int)

display(
    stock[
        [
            "Date",
            "Days_To_Earnings",
            "Earnings_Within_5D",
            "Earnings_Within_10D",
            "Earnings_Within_20D"
        ]
    ].tail(30)
)
```

```python
# ============================================
# Step 57 - Feature Selection
# ============================================

features = [
    # 가격
    "Open",
    "High",
    "Low",
    "Close",

    # 거래량
    "Volume",
    "Volume_MA20",
    "Volume_Change",

    # 기술적 지표
    "Return_1D",
    "MA5",
    "MA20",
    "MA60",
    "Volatility_20D",
    "RSI14",

    # 시장
    "SPY_Return",
    "QQQ_Return",

    # SEC 공시
    "Filing_Count",
    "Filing_8_K",
    "Filing_10_Q",
    "Filing_10_K",
    "Filing_S_1",
    "Filing_S_3",
    "Filing_DEF_14A",
    "Daily_Event_Score",

    # 재무
    "Revenue",
    "Net_Income",
    "Assets",
    "Liabilities",
    "Stockholders_Equity",
    "Cash",
    "Operating_Cash_Flow",

    # 가치평가
    "Market_Cap",
    "PS_Ratio",
    "PB_Ratio",

    # 재무비율
    "Debt_Ratio",
    "Cash_to_Liabilities",
    "ROA",
    "ROE",
    "Revenue_Growth",

    # 공매도
    "Short_Interest",
    "Short_Interest_Ratio",
    "Short_Volume",
    "Short_Volume_Ratio",
    "Short_Interest_to_Volume",
    "Short_Ratio_Change",

    # 이벤트
    "Days_To_Earnings",
    "Earnings_Within_5D",
    "Earnings_Within_10D",
    "Earnings_Within_20D"
]

print("Feature 개수:", len(features))
```

```python
# ============================================
# Step 58 - Create Model Dataset
# ============================================

# features 리스트 내의 'Filing_DEF_14A' 오기를 실제 컬럼명 'Filing_DEF 14A'로 보정
corrected_features = [f if f != "Filing_DEF_14A" else "Filing_DEF 14A" for f in features]

model_df = stock[
    ["Date"] + corrected_features + ["Target", "Next_Return"]
].copy()

print("모델 데이터 크기:", model_df.shape)

display(model_df.head())
```

```python
# ============================================
# Step 60 - Sort Chronologically
# ============================================

model_df = model_df.sort_values("Date")

model_df.reset_index(drop=True, inplace=True)

display(model_df.head())
display(model_df.tail())
```

```python
# ============================================
# Step 61 - Time Series Split
# ============================================

n = len(model_df)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

train_df = model_df.iloc[:train_end].copy()
val_df = model_df.iloc[train_end:val_end].copy()
test_df = model_df.iloc[val_end:].copy()

print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)

print("\nTrain 기간:")
print(train_df["Date"].min(), "~", train_df["Date"].max())

print("\nValidation 기간:")
print(val_df["Date"].min(), "~", val_df["Date"].max())

print("\nTest 기간:")
print(test_df["Date"].min(), "~", test_df["Date"].max())
```

```python
# ============================================
# Step 62 - X / y Split
# ============================================

# 원본 features 대신 공백 보정이 완료된 corrected_features를 사용하여 데이터 분할
X_train = train_df[corrected_features].copy()
y_train = train_df["Target"].copy()

X_val = val_df[corrected_features].copy()
y_val = val_df["Target"].copy()

X_test = test_df[corrected_features].copy()
y_test = test_df["Target"].copy()

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_val:", X_val.shape)
print("y_val:", y_val.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
# ============================================
# Step 62 - X / y Split
# ============================================

# 오기가 수정된 corrected_features를 사용하여 데이터 분할을 진행합니다.
X_train = train_df[corrected_features].copy()
y_train = train_df["Target"].copy()

X_val = val_df[corrected_features].copy()
y_val = val_df["Target"].copy()

X_test = test_df[corrected_features].copy()
y_test = test_df["Target"].copy()

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_val:", X_val.shape)
print("y_val:", y_val.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
# ============================================
# Step 63 & 64 - Feature Scaling & ANN Model
# ============================================

import numpy as np
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# 1. Feature Scaling 진행 (무한대 값 처리 추가)
scaler = StandardScaler()

# 모든 무한대(inf, -inf) 값을 NaN으로 변환한 뒤, NaN을 0으로 채워줍니다.
X_train_clean = X_train.replace([np.inf, -np.inf], np.nan).fillna(0)
X_val_clean = X_val.replace([np.inf, -np.inf], np.nan).fillna(0)
X_test_clean = X_test.replace([np.inf, -np.inf], np.nan).fillna(0)

X_train_scaled = scaler.fit_transform(X_train_clean)
X_val_scaled = scaler.transform(X_val_clean)
X_test_scaled = scaler.transform(X_test_clean)

print("데이터 스케일링 완료!")
print("X_train_scaled shape:", X_train_scaled.shape)

# 2. ANN 모델 정의
model = Sequential([
    Dense(
        128,
        activation="relu",
        input_shape=(X_train_scaled.shape[1],)
    ),

    Dropout(0.20),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.20),

    Dense(
        32,
        activation="relu"
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        1,
        activation="sigmoid"
    )
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
# ============================================
# Step 65 - Train ANN
# ============================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True
)

history = model.fit(
    X_train_scaled,
    y_train,

    validation_data=(
        X_val_scaled,
        y_val
    ),

    epochs=200,
    batch_size=32,

    callbacks=[
        early_stopping
    ],

    verbose=1
)
```

```python
# ============================================
# Step 66 - Training History
# ============================================

plt.figure(figsize=(10,5))

plt.plot(
    history.history["loss"],
    label="Train Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("ANN Training Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()

plt.show()
```

```python
# ============================================
# Step 67 - Test Prediction
# ============================================

test_probability = model.predict(
    X_test_scaled
).flatten()

test_prediction = (
    test_probability >= 0.5
).astype(int)

print("예측 완료")
```

```python
# ============================================
# Step 80 - Permutation Importance (Fixed Wrapper with fit)
# ============================================

from sklearn.inspection import permutation_importance
import pandas as pd

# Keras 모델의 확률 출력을 이진 클래스(0 또는 1)로 변환해주는 래퍼 클래스 정의
# scikit-learn 검증 통과를 위해 빈 fit 메서드를 추가합니다.
class KerasClassifierWrapper:
    def __init__(self, keras_model):
        self.keras_model = keras_model
        self.classes_ = [0, 1]

    def fit(self, X, y=None):
        # 이미 학습된 모델을 사용하므로 아무 작업도 하지 않고 자기 자신을 반환합니다.
        return self

    def predict(self, X):
        proba = self.keras_model.predict(X, verbose=0).flatten()
        return (proba >= 0.5).astype(int)

# 래퍼 적용
wrapped_model = KerasClassifierWrapper(model)

# 순열 중요도 계산
perm = permutation_importance(
    wrapped_model,
    X_test_scaled,
    y_test,
    n_repeats=10,
    random_state=42,
    scoring="accuracy"
)

importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": perm.importances_mean
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
)

display(importance_df.head(20))
```

```python
# ============================================
# Step 84 - Prediction Function (Fixed Feature Columns & Handling NaNs)
# ============================================

def predict_atch(row):
    import pandas as pd
    import numpy as np

    # 입력받은 row(Series 또는 dict-like)를 DataFrame 형태로 구성
    row_df = pd.DataFrame([row[corrected_features]])

    # 무한대 값 및 결측치 처리 (학습 데이터 전처리 스펙과 동일하게 적용)
    row_clean = row_df.replace([np.inf, -np.inf], np.nan).fillna(0.0)
    X = row_clean.values.astype(np.float64)

    # 1. 상승/하락 방향 예측
    X_scaled = scaler.transform(X)
    direction_probability = model.predict(X_scaled, verbose=0)[0][0]

    # 2. 예상 수익률 예측
    X_return_scaled = reg_scaler.transform(X)
    expected_return = reg_model.predict(X_return_scaled, verbose=0)[0][0]

    return {
        "상승확률": direction_probability,
        "예상수익률": expected_return
    }
```
